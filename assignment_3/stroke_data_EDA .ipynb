```python
corr_matrix = df_corr.corr(numeric_only=True)
```

```python
plt.figure(figsize=(12, 8))
sns.heatmap(corr_matrix, annot=True)
plt.title('Correlation Matrix')
plt.show()
```

### Final EDA Insights

1. The dataset is heavily skewed, with only **4.87%** of patients having experienced a stroke compared to **95.12%** who have not.

2. Based on correlation analysis, age is the single strongest numerical predictor of stroke risk. Secondary risk amplifiers include pre-existing medical conditions like hypertension, heart_disease, and elevated avg_glucose_level.

3. Surprisingly, bmi and gender exhibit virtually zero linear correlation with stroke occurrence in this dataset. While ever_married appears slightly correlated with stroke, it is actually a confounding variable driven by age (older individuals are highly likely to be both married and at a higher risk for stroke).

4. While we successfully median-imputed the 201 missing bmi values, the smoking_status column presents a larger data quality challenge. With 1,544 "Unknown" entries (nearly 30% of the dataset), we cannot drop these rows without losing valuable data.
